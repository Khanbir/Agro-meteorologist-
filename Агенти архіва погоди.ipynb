```python
%pip install meteostat pandas

from datetime import datetime, timedelta
from meteostat import Point, Daily
import pandas as pd

# === Агент 1: Збір погодних даних ===
class WeatherCollector:
    def __init__(self, latitude, longitude, days=14):
        self.location = Point(latitude, longitude)
        self.days = days
        self.data = None

    def fetch_weather_data(self):
        end = datetime.now()
        start = end - timedelta(days=self.days)
        try:
            data = Daily(self.location, start, end)
            data = data.fetch()
            if data.empty:
                print("[Agent 1] ❌ Дані порожні. Можливо, поблизу немає метеостанцій.")
            else:
                print(f"[Agent 1] ✅ Отримано {len(data)} днів даних")
            self.data = data.reset_index()
        except Exception as e:
            print(f"[Agent 1] ❌ Помилка при отриманні даних: {e}")

    def export_csv(self, path="weather_data.csv"):
        if self.data is not None and not self.data.empty:
            self.data.to_csv(path, index=False)
            print(f"[Agent 1] ✅ Дані збережено в {path}")
        else:
            print("[Agent 1] ⚠ Дані відсутні. CSV не створено.")

# === Агент 2: Аналіз і прогноз внесення добрив ===
class FertilizationPlanner:
    def __init__(self, csv_path="weather_data.csv"):
        try:
            self.df = pd.read_csv(csv_path)
            print(f"[Agent 2] ✅ CSV-файл завантажено. Рядків: {len(self.df)}")
        except Exception as e:
            print(f"[Agent 2] ❌ Помилка при зчитуванні CSV: {e}")
            self.df = pd.DataFrame()

    def analyze(self):
        if self.df.empty:
            print("[Agent 2] ❌ Дані відсутні для аналізу.")
            return

        if 'prcp' not in self.df.columns or 'tavg' not in self.df.columns:
            print("[Agent 2] ❌ Відсутні необхідні колонки 'prcp' та 'tavg'")
            print(f"[Agent 2] Наявні колонки: {list(self.df.columns)}")
            return

        dry_days = self.df[self.df['prcp'] < 1.0]
        suitable_days = dry_days[(dry_days['tavg'] >= 8) & (dry_days['tavg'] <= 20)]

        if not suitable_days.empty:
            print(f"[Agent 2] ✅ Рекомендовано внести добрива {suitable_days.iloc[0]['time']}")
        else:
            print("[Agent 2] ⛔ Не знайдено підходящих днів (t 8–20°C та без опадів)")

# === Головна логіка ===
def main():
    latitude = 47.9050
    longitude = 33.6540

    print("\n▶ Запуск агента 1: Збір історичних погодних даних...")
    collector = WeatherCollector(latitude, longitude)
    collector.fetch_weather_data()
    collector.export_csv()

    print("\n▶ Запуск агента 2: Аналіз і прогноз внесення добрив...")
    planner = FertilizationPlanner()
    planner.analyze()

if __name__ == "__main__":
    main()
```
